```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Configuração dos caminhos do projeto

PROJECT_ROOT = Path.cwd().parent

PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
FIGURES_DIR = PROJECT_ROOT / "results" / "figures"
TABLES_DIR = PROJECT_ROOT / "results" / "tables"

PROCESSED_FILE = PROCESSED_DIR / "wine_quality_clean.csv"

FIGURES_DIR.mkdir(parents=True, exist_ok=True)
TABLES_DIR.mkdir(parents=True, exist_ok=True)
```

```python
# Carregamento da base limpa
wine_df = pd.read_csv(PROCESSED_FILE)
wine_df.head()
```

```python
# Boxplot da variável Fixed Acidity
# Usado para visualizar a distribuição, mediana e possíveis outliers da acidez fixa

plt.figure(figsize=(8, 5))
plt.boxplot(wine_df["fixed acidity"])
plt.title("Boxplot da Acidez Fixa")
plt.xlabel("Acidez Fixa")
plt.ylabel("Valores")

plt.tight_layout()
plt.savefig(FIGURES_DIR / "boxplot_fixed_acidity.png", bbox_inches="tight")
plt.show()

wine_df["fixed acidity"].min(), wine_df["fixed acidity"].max()
```
